```python
# Simulador de IA Agrícola: "AgroBot 3000"
# Diseñado para ayudar a un agricultor a cuidar sus cultivos

print("=========================================")
print("🤖 ¡Bienvenido a AgroBot 3000! 🌾")
print("Tu asistente de Inteligencia Artificial")
print("=========================================")

# Bucle principal para mantener el menú interactivo
while True:
    print("\n--- MENÚ PRINCIPAL DEL AGRICULTOR ---")
    print("1. Analizar estado de la tierra")
    print("2. Sistema de riego inteligente")
    print("3. Alerta de plagas y clima")
    print("4. Apagar AgroBot 3000")

    opcion = input("\nSelecciona una opción (1-4): ")

    # --- OPCIÓN 1: ANALIZAR LA TIERRA ---
    if opcion == "1":
        print("\n[IA] Analizando sensores de humedad...")
        humedad = input("¿El suelo está 'seco', 'perfecto' o 'inundado'?: ")

        if humedad == "seco":
            print("🚨 Alerta: La tierra necesita nutrientes urgentes y agua.")
        elif humedad == "perfecto":
            print("✨ ¡Excelente! El suelo tiene las condiciones óptimas para el cultivo.")
        elif humedad == "inundado":
            print("⚠️ Cuidado: Demasiada agua. Activa los canales de drenaje.")
        else:
            print("❌ Entrada no válida. Por favor, escribe 'seco', 'perfecto' o 'inundado'.")

    # --- OPCIÓN 2: RIEGO INTELIGENTE ---
    elif opcion == "2":
        print("\n[IA] Evaluando necesidad de riego...")
        clima = input("¿Cómo está el clima actual? ('lluvia' o 'sol'): ")

        if clima == "lluvia":
            print("🌧️ IA decide: Riego APAGADO para ahorrar agua y evitar inundaciones.")
        elif clima == "sol":
            print("☀️ IA decide: Riego ACTIVADO. Regando los cultivos de forma eficiente.")
        else:
            print("❌ Clima no reconocido. Intenta de nuevo.")

    # --- OPCIÓN 3: ALERTA DE PLAGAS ---
    elif opcion == "3":
        print("\n[IA] Escaneando hojas con cámara simulada...")
        color_hojas = input("¿De qué color son las hojas? ('verdes' o 'amarillas'): ")

        if color_hojas == "verdes":
            print("🌿 Cultivo sano. No se detectan anomalías ni plagas.")
        elif color_hojas == "amarillas":
            print("🐛 ¡Alerta de Plaga o Falta de Nitrógeno! Se recomienda revisar la zona sur.")
        else:
            print("❌ Color no válido. Escribe 'verdes' o 'amarillas'.")

    # --- OPCIÓN 4: SALIR DEL PROGRAMA ---
    elif opcion == "4":
        print("\n🤖 Guardando datos en la nube... AgroBot 3000 se ha apagado. ¡Hasta pronto, agricultor!")
        break # Rompe el ciclo while True para terminar el programa

    # --- VALIDACIÓN DEL MENÚ ---
    else:
        print("⚠ Opcón no válida. Por favor, elige un número del 1 al 4.")

    print("\n-----------------------------------------")
```
